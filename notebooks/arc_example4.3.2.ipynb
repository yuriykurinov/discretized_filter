```python
theta_labels = ['$\\theta_t = e_1$', '$\\theta_t = e_2$']

theta_colors = ['lightgray', 'k']

abc = ['а', 'б', 'в', 'г']

M = 2
```

```python
def plot_theta(theta, t, theta_opt_list, t_nets, options, t_net_filtering, abc=abc):

    fig, ax = plt.subplots(figsize=((1 + np.sqrt(5))/2 * 8, 4), layout='constrained')

    hts = [t_net[1] - t_net[0] for t_net in t_nets]

    est_lines = []
    # line1, = ax.step(
    #     [0] + list(t), 
    #     [(theta[0] == 0)] + list(theta == 0), 
    #     lw=1.5,
    #     color='k'
    # )
    from discretized_filter.utils.grids import to_discrete
    dtheta = to_discrete(theta, t, 10, 0.001)
    line1, = ax.plot(
        t_net_filtering[:-1],
        (dtheta == 0),
        lw=0.7, color='gainsboro', marker='s', markersize=6
    )
    for t_net, theta_opt, option in zip(t_nets, theta_opt_list, options):
        line4, = ax.plot(t_net, theta_opt[:, 0], **option)
        est_lines.append(line4)

    ax.set(
        xlim=(0, 10),
        ylim=(-0.1, 1.1)
    )
    #ax.set_xlabel('({})'.format(abc[]), fontsize=13)
    ax.set_ylabel('$\\theta_t^1$', fontsize=20, rotation='horizontal')
    ax.yaxis.set_label_coords(-0.03, 0.4)
    ax.tick_params(labelsize=20)
    ax.set_xticks(
        [0, 2, 4, 6, 8, 10], 
        ['0', '2', '4', '6', '8', '$t$']
    )
    ax.set_yticks(ticks=[0, 1])
    ax.spines["bottom"].set_color("k")
    ax.spines["top"].set_color("k")
    ax.spines["left"].set_color("k")
    ax.spines["right"].set_color("k")

    fig.legend(
        fontsize=18,
        handles=[line1] + est_lines,
        labels=['$\\theta_t^i$'] + [
            f'$\\overline{{\\theta}}_t^1$ при $\\Delta_t = {{{str(ht).replace('.', ',\\!')}}}$' for ht in hts
        ],
        loc='outside lower center',
        ncol=5,
        framealpha=1,
        frameon=False
    )

    return fig, ax
```

```python
from math import ceil

t_nets = [
    np.array([i*step for i in range(ceil(10 / step)+1)])
    for step in [0.5, 0.1, 0.01, 0.001,]
]
# 0.01
```
